### Problem 1

Let $f:\mathbb{R}^{2}\rightarrow\mathbb{R}$ be the function given by:
$$f(x,y)=x^{2}+y$$

Also define the subset of $\mathbb{R}^{2}$:
$$C=\{(x,y)\in\mathbb{R}^{2} \mid x\le1, y\le1, x+y\ge1\}$$

**(a)** Show that $f$ is a convex function.

**(b)** Show that $f$ is not strictly convex.

Consider the optimization problem (1):
$$\min ~ f(x,y)$$
$$(x,y)\in C$$

**(c)** Is the optimization problem (1) a convex optimization problem? If so, why?

**(d)** Express the set $U=\mathbb{R}^{2}\setminus C$ using strict inequalities (i.e., $>$ and $<$) in $x$ and $y$.
Is $U$ an open subset?

**(e)** Explain, by a simple reference to the curriculum without calculations, why there must exist an optimal solution to (1).

**(f)** Solve the optimization problem (1) by only using the KKT conditions for (1).
Is there more than one solution to the KKT conditions for (1)?

By [Theorem 8.23](imo25.pdf#section.8.5) $f$ is convex if and only if the Hessian matrix is positive semidefinite. Calculating the hessian matrix yields the result of $ H = \left(\begin{array}{rr}
2 & 0 \\
0 & 0
\end{array}\right)$ which by [Section 3.7.2](imo25.pdf#subsection.3.7.2) is positive definite if $f_H(v) \geq 0$ for all non zero values of $v$. $f_H=2x^2$ for this particular hessian matrix which is nonnegative for all real values of $x$

```python
var('x y')
f(x,y) = x^2+y
h : Matrix = f.hessian()
v = matrix([[x], [y]])
v.transpose()*h*v
```

To show that the function $f$ is not convex we need to find points which break the [Definition 6.1](imo25.pdf#section.6.1) such points are for example $(0, 0)$ and $(1,0)$  

```python
var('x1 y1 x2 y2 t')
LHS = f((1-t)*x1 + t*x2, (1-t)*y1 + t*y2)
RHS = (1-t)*f(x1, y1) + t*f(x2, y2)
equation = LHS == RHS
solve([equation, 0<t, t<1], x1, y1, x2, y2, t)
```

According to [Chapter 9](imo25.pdf#chapter.9) Convex optimization problem is one where we minimize function $f$ over subset $C$ where $f$ is a convex function and $C$ is a convex subset.
By rearranging the third condition in set $C$ we can write it as $C=\{(x,y)\in\mathbb{R}^{2} \mid x\le1, y\le1, 1-x-y\leq 0\}$. Since those are all in the form of $g_i\leq a$ and by the result of [Exercise 4.31](imo25.pdf#section.4.4) then by [Lemma 4.27](imo25.pdf#section.4.3) all of these conditions form convex subsets and by the result of [Exercise 4.23](imo25.pdf#section.4.3) the set $C$ is convex.

And since $C$ is convex set and $f(x,y)$ is convex function the stated problem is a convex optimization problem

$\mathbb{R} \setminus C$ can be written as
$$U=\{(x,y)\in\mathbb{R}^{2} \mid x > 1 \lor y > 1 \lor x + y < 1 \}$$
where set $U$ is open by [Definition 5.38](imo25.pdf#subsection.5.6.2)

By [Remark 5.59](imo25.pdf#subsection.5.7.2) the function $f$ is continuous and since set $C$ is compact (by being bounded and closed) then by [Theorem 5.66](imo25.pdf#section.5.8) the problem has optimal solution

For the minimization problem we get the following KKT conditions:
$$
\begin{aligned}
\lambda_{0} &\geq 0 \\
\lambda_{1} &\geq 0 \\
\lambda_{2} &\geq 0 \\
x - 1 &\leq 0 \\
y - 1 &\leq 0 \\
-x - y + 1 &\leq 0 \\
\lambda_{0} {\left(x - 1\right)} &= 0 \\
\lambda_{1} {\left(y - 1\right)} &= 0 \\
\lambda_{2} {\left(1 - x - y\right)} &= 0 \\
\lambda_{0} - \lambda_{2} + 2 \, x &= 0 \\
\lambda_{1} - \lambda_{2} + 1 &= 0 \\
\end{aligned}
$$
From this follow that
$$
\begin{aligned}
\lambda_3 = 1 + \lambda_2 &\Rightarrow \\
\lambda_3 \geq 1 \land x + y = 1 \land 2x = 1 - \lambda_1 + \lambda_2
\end{aligned}
$$
If $\lambda_1 > 0$ then $x=1$. Since $x+y=1$ then $y=0$, since $y<1$ then $\lambda_2=0$. Substitute into $2x = 1 - \lambda_1 + \lambda_2$ and $2=1-\lambda_1 \Rightarrow \lambda_1 = -1$ which contradicts $\lambda_1 \geq 0$

If $\lambda_2 > 0$ then $y=1$. Since $x+y=1$ then $x=0$, since $x<1$ then $\lambda_1=0$. Substitute into $2x = 1 - \lambda_1 + \lambda_2$ and $0=1+\lambda_2 \Rightarrow \lambda_2 = -1$ which contradicts $\lambda_2 \geq 0$

If $\lambda_1 = 0 \land \lambda_2 = 0$. Substitue into $2x = 1 - \lambda_1 + \lambda_2$ then $x=0.5$ and since $x+y=1$ then $y=0.5$ and all conditions hold true. Therefore the only result to the KKT conditions is $x=0.5$ and $y=0.5$

### Problem 2

**(d)** Does the function $g(x,y)$ have a global minimum?

The hessian is $\left(\begin{array}{rr}
(2+4x^{2})e^{\left(x^{2}+y\right)} & 2 \, x e^{\left(x^{2} + y\right)} \\
2 \, x e^{\left(x^{2} + y\right)} & e^{\left(x^{2} + y\right)}

4 \, x^{2} + 2 & 2 \, x \\
2 \, x & 1
\end{array}\right)$ since $e^{\left(x^{2}+y\right)}>0$ we check the matrix using the criteria from above which shows that the matrix is positive definite. Then by [Theorem 8.23](imo25.pdf#section.8.5) the function $g$ is strictly convex.

```python
h : Matrix = (g.hessian()/e^(x^2+y)).factor()
print(latex(h))
show(h[0,0])
show((h[0,0]*h[1,1]-h[1,0]*h[1,0]))
```

The function is defined as $g(x,y) = e^{x^2+y}$ for all $(x,y) \in \mathbb{R}^2$.

1.  **Strict Positivity:** The range of the exponential function $e^z$ is $(0, \infty)$. Therefore, for any real numbers $x$ and $y$:
    $$g(x,y) = e^{x^2+y} > 0$$
    The function never takes a value of 0 or less.

2.  **Infimum is 0:** We can show that the values of $g(x,y)$ get arbitrarily close to 0. Consider the limit as $y$ approaches $-\infty$ while keeping $x$ fixed (e.g., $x=0$):
    $$\lim_{y \to -\infty} g(0, y) = \lim_{y \to -\infty} e^{0^2+y} = \lim_{y \to -\infty} e^y = 0$$

**Conclusion:**
Since the infimum of the function is 0, but the function never actually attains the value 0 (it is strictly positive everywhere), there is no point $(x^*, y^*)$ where the minimum is reached. Therefore, $g(x,y)$ does not have a global minimum.

Since the set is compact as shown in previous problem and the function is strictly convex the minimization problem given has a solution by [Theorem 5.66](imo25.pdf#section.5.8) and by [Theorem 6.13](imo25.pdf#section.6.2) the solution is unique

We want to solve:
$$\min_{(x,y) \in C} g(x,y) = e^{x^2+y}$$
where $C = \{(x,y) \in \mathbb{R}^2 \mid x \le 1, y \le 1, x+y \ge 1\}$.

**Simplification:**
Since the exponential function $z \mapsto e^z$ is strictly increasing, minimizing $g(x,y)$ is equivalent to minimizing the exponent:
$$h(x,y) = x^2 + y$$

**1. Interior Critical Points:**
We calculate the gradient of $h$:
$$\nabla h(x,y) = \begin{pmatrix} 2x \\ 1 \end{pmatrix}$$
Since the second component is always $1 \neq 0$, $\nabla h(x,y)$ is never the zero vector. Thus, there are no stationary points in the interior of $C$. The minimum must lie on the boundary.

**2. Boundary Analysis:**
The set $C$ is a triangle with vertices $(1,0)$, $(0,1)$, and $(1,1)$. We analyze $h(x,y)$ on each edge.

* **Edge 1 ($y=1, 0 \le x \le 1$):**
    $$f_1(x) = h(x,1) = x^2 + 1$$
    This is increasing for $x \in [0,1]$. The minimum on this edge is at $x=0$, giving point $(0,1)$ with value $h(0,1) = 1$.

* **Edge 2 ($x=1, 0 \le y \le 1$):**
    $$f_2(y) = h(1,y) = 1 + y$$
    This is increasing for $y \in [0,1]$. The minimum on this edge is at $y=0$, giving point $(1,0)$ with value $h(1,0) = 1$.

* **Edge 3 ($x+y=1 \implies y=1-x, 0 \le x \le 1$):**
    Substitute $y = 1-x$ into $h$:
    $$f_3(x) = x^2 + (1-x) = x^2 - x + 1$$
    Find the critical point by differentiating with respect to $x$:
    $$f_3'(x) = 2x - 1 = 0 \implies x = \frac{1}{2}$$
    Since $x=1/2$ is in the interval $[0,1]$, we evaluate $h$ at this point (where $y = 1 - 0.5 = 0.5$):
    $$h\left(\frac{1}{2}, \frac{1}{2}\right) = \left(\frac{1}{2}\right)^2 - \frac{1}{2} + 1 = \frac{1}{4} - \frac{2}{4} + \frac{4}{4} = \frac{3}{4}$$

**3. Global Minimum:**
We compare the candidate minimum values found on the boundaries:
* $h(0,1) = 1$
* $h(1,0) = 1$
* $h(1/2, 1/2) = 0.75$

The lowest value is $0.75$. Therefore, the minimum value of the original function $g(x,y)$ is:
$$g\left(\frac{1}{2}, \frac{1}{2}\right) = e^{3/4}$$

**Solution:** The unique solution is $(x,y) = \left(\frac{1}{2}, \frac{1}{2}\right)$.

Minimizing $g(x,y) = e^{x+y}$ is equivalent to minimizing the linear function:
$$k(x,y) = x + y$$
subject to the constraints of $C$.

One of the constraints defining the set $C$ is:
$$x + y \ge 1$$
This implies that for any point $(x,y) \in C$, the value of the objective function $k(x,y)$ must be at least 1.

The boundary edge connecting $(1,0)$ and $(0,1)$ is defined exactly by the line $x+y=1$. For every point on this line segment, the objective function attains the value $k(x,y) = 1$. Since $k(x,y) \ge 1$ everywhere else in $C$, the minimum value is indeed 1.

**Conclusion:**
Since the minimum value is achieved at every point on the line segment between $(1,0)$ and $(0,1)$ (e.g., $(1,0)$, $(0,1)$, $(0.5, 0.5)$, etc.), there are infinitely many solutions.
Therefore, the solution is **not unique**.

### Problem 3

Consider the system of linear equations

$$
\begin{alignat}{4}
x & {}+{} & y & {}+{} & z & {}={} & 1 \\
x & {}+{} & 2y & {}+{} & 3z & {}={} & 1 \\
  &       & y & {}+{} & 2z & {}={} & 1
\end{alignat} \tag{3}
$$

in the three variables $x, y$, and $z$.

(a) Let

$$
v = \begin{pmatrix} x \\ y \\ z \end{pmatrix}.
$$

Find a $3 \times 3$ matrix $A$ and a $3 \times 1$ matrix $b$ such that the system of equations (3) is expressed as $Av = b$.

(b) Show that there exists no $v \in \mathbb{R}^3$ such that $Av = b$ (the system of equations (3) cannot be solved).

(c) Show that the matrix $A$ is not invertible.

(d) Solve the optimization problem

$$
\min_{v \in \mathbb{R}^3} \quad |b - Av|^2.
$$

It is permitted to use a computer for matrix calculations here. Is the solution to the optimization problem unique?

$\left(\begin{array}{rrr}
1 & 1 & 1 \\
1 & 2 & 3 \\
0 & 1 & 2
\end{array}\right) \left(\begin{array}{r}
x \\
y \\
z
\end{array}\right)= \left(\begin{array}{r}
1 \\
1 \\
1
\end{array}\right)$

```python
var('x y z')
A = matrix([[1,1,1], [1,2,3], [0,1,2]])
v = matrix([[x],[y],[z]])
b = matrix([[1],[1],[1]])
print(latex(A), latex(v), latex(b))
```

$$
\begin{aligned}
y + 2z =1 \Rightarrow y = 1 - 2z &\Rightarrow \\
x+2-4z+3z=1 \Rightarrow x = z - 1 &\Rightarrow \\
z-1+1-2z+z=1 \Rightarrow 0=1&\\
\end{aligned}
$$
which is false meaning there is no solution to this system of linear equations

The matrix is not invertible because the rows are lineary dependent. Specifically $R_2=R_1+R_3$

```python
lhs = (A.transpose()*A)*v
rhs = A.transpose()*b
show(rhs, lhs)
eqs = []
for i in range(2):
    eqs.append(lhs[i,0]==rhs[i,0])
solve(eqs, [x,y,z])
```

The solution to the minimization problem is $x=z$, $y=\frac{2}{3}-2z$
