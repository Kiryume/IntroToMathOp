#### Invertability of Matrix and the inverse matrix

**1. The $2 \times 2$ Case**
For a matrix $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$, invertibility is determined by the determinant-like quantity $D = ad - bc$ **[Exercise 3.24](imo25.pdf#section.3.5)**.
* **Condition:** $A$ is invertible if and only if $D \neq 0$ **[Exercise 3.24](imo25.pdf#section.3.5)**.
* **Formula:** If invertible, the inverse is given by $A^{-1} = \frac{1}{D} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$ **[Exercise 3.24](imo25.pdf#section.3.5)**.

**2. General Method ($n \times n$)**
An $n \times n$ matrix $A$ is invertible if there exists a matrix $B$ such that $AB = BA = I_n$ **[Definition 3.22](imo25.pdf#section.3.5)**. Finding the inverse corresponds to solving $n$ systems of linear equations. To find the columns $B^j$ of the inverse matrix $B$, we must solve:
$$AB^j = I^j$$
for each column $j=1, \dots, n$ of the identity matrix $I$ **[Section 3.5.1](imo25.pdf#subsection.3.5.1)**.

**3. Other Criteria**
* **Zero Rows/Columns:** A quadratic matrix containing a row or column consisting entirely of zeros cannot be invertible **[Exercise 3.23](imo25.pdf#section.3.5)**.
* **Transpose:** A quadratic matrix $A$ is invertible if and only if its transpose $A^\top$ is invertible **[Exercise 3.34](imo25.pdf#section.3.6)**.

***note** use the code below to get the linear equations needed to find the inverse matrix (replace the `show` function with `print(latex(...))` to get the latex code)*

```python
# 1. Define the Matrix
# Replace with your matrix values
A = matrix(QQ, [[1, 2, 3], [1, 1, 4], [0, 6, 0]]) 

# 3. Find Inverse by Solving Linear Equations (General Method)
n = A.nrows()
I = identity_matrix(n)

print("Solving systems AB = I:")
B = []
for i in range(n):
    row = []
    for j in range(n):
        row.append(var('b_{}{}'.format(i+1, j+1)))
    B.append(row)
B = matrix(B)
C = A * B
equations = []
for i in range(n):
    for j in range(n):
        equations.append(C[i, j] == I[i, j])
for eq in equations:
    # print(latex(eq))
    show(eq)
solutions = solve(equations, B.list())
if len(solutions) == 0:
    print("Matrix is not invertible")
else:
    B_sol = B.subs(solutions[0])
    print("Inverse Matrix B found by solving equations:")
    # print(latex(B_sol))
    show(B_sol)
    print("B = A^(-1) (should be identity):")
    show(B_sol*A)
```
