# Problem 3 Solution

This notebook solves Problem 3 from the May 2025 exam using the syllabus provided in `imo25.pdf`.

**Problem Statement:**
Let $C \subseteq \mathbb{R}^3$ be the set of points $(x,y,z) \in \mathbb{R}^3$ that satisfy the following inequalities:
1. $x+y-z \le 1$
2. $-x+2y+z \le 2$
3. $x-y+z \le 3$
4. $-x-y-z \le 4$

## Part (a): Fourier-Motzkin Elimination of $z$

We use **Fourier-Motzkin elimination** as described in **[Section 4.5](imo25.pdf#section.4.5)** to eliminate the variable $z$. 

First, we isolate $z$ in each of the four inequalities to obtain lower and upper bounds for $z$:

From (1): $z \ge x + y - 1$  
From (2): $z \le 2 + x - 2y$  
From (3): $z \le 3 - x + y$  
From (4): $z \ge -x - y - 4$  

This gives us two lower bounds ($L_1, L_2$) and two upper bounds ($U_1, U_2$):

* $L_1(x,y) = x + y - 1$
* $L_2(x,y) = -x - y - 4$
* $U_1(x,y) = 2 + x - 2y$
* $U_2(x,y) = 3 - x + y$

To eliminate $z$, we form inequalities $L_i \le U_j$ for all combinations of $i$ and $j$.

1.  **$L_1 \le U_1$:**
    $$x + y - 1 \le 2 + x - 2y \iff 3y \le 3 \iff y \le 1$$

2.  **$L_1 \le U_2$:**
    $$x + y - 1 \le 3 - x + y \iff 2x \le 4 \iff x \le 2$$

3.  **$L_2 \le U_1$:**
    $$-x - y - 4 \le 2 + x - 2y \iff -2x + y \le 6 \iff y \le 2x + 6$$

4.  **$L_2 \le U_2$:**
    $$-x - y - 4 \le 3 - x + y \iff -2y \le 7 \iff y \ge -3.5$$

**Result of eliminating $z$:**
The projection of $C$ onto the $xy$-plane is defined by:
$$ \begin{cases} y \le 1 \\ x \le 2 \\ y \le 2x + 6 \\ y \ge -3.5 \end{cases} $$

```python
load("math_utils.py")
var('x y z')

# The system from Problem 3
# 1. x + y - z <= 1
# 2. -x + 2*y + z <= 2
# 3. x - y + z <= 3
# 4. -x - y - z <= 4
system = [
    x + y - z <= 1,
    -x + 2*y + z <= 2,
    x - y + z <= 3,
    -x - y - z <= 4
]

print("=== ORIGINAL SYSTEM ===")
for i in system:
    print(i)
print("\n")

# Eliminate z first, then y (as done in the exam solution)
final_system = solve_fourier_motzkin(system, [z, y])

print("=== FINAL SYSTEM (after elimination) ===")
for ineq in final_system:
    # Simplify the output for readability
    print(ineq.simplify())
    print(solve([ineq], x,y,z))
```

## Part (b): Elimination of $y$

Next, we eliminate $y$ from the inequalities obtained in Part (a).

We isolate $y$ to identify lower and upper bounds:
* Upper bounds ($U_y$): $y \le 1$ and $y \le 2x + 6$
* Lower bound ($L_y$): $y \ge -3.5$

We compare the lower bound to each upper bound:

1.  **$-3.5 \le 1$:**
    This is always true ($0 \le 4.5$) and provides no constraints on $x$.

2.  **$-3.5 \le 2x + 6$:**
    $$-9.5 \le 2x \iff x \ge -4.75$$

We also retain the independent constraint on $x$ from Part (a): $x \le 2$.

**Result of eliminating $y$:**
$$ -4.75 \le x \le 2 $$

## Part (c): Showing $C$ is Bounded

We use the definition of a **bounded subset** from **[Definition 5.29](imo25.pdf#subsection.5.6.1)**. A subset is bounded if it is contained in a ball of finite radius, which essentially means all coordinates must be bounded.

1.  **Bounds for $x$:**
    From Part (b), we established that:
    $$ x \in [-4.75, 2] $$
    Thus, $|x|$ is bounded by $M_x = 4.75$.

2.  **Bounds for $y$:**
    From Part (a), we have $y \ge -3.5$ and $y \le 1$ (and $y \le 2x+6$).
    Since $y$ is bounded below by $-3.5$ and above by $1$, it lies in a finite interval:
    $$ y \in [-3.5, 1] $$
    Thus, $|y|$ is bounded by $M_y = 3.5$.

3.  **Bounds for $z$:**
    From Part (a), $z$ is bounded by linear combinations of $x$ and $y$:
    $$ \max(x+y-1, -x-y-4) \le z \le \min(2+x-2y, 3-x+y) $$
    Since $x$ and $y$ are bounded, these linear combinations are finite. For instance:
    $$ |z| \le \max(|x|+|y|+1, |x|+|y|+4, 2+|x|+2|y|, 3+|x|+|y|) $$
    Given $M_x$ and $M_y$, there exists a constant $M_z$ such that $|z| \le M_z$.

**Conclusion:**
Since coordinates $x, y,$ and $z$ are all bounded, the set $C$ is bounded.

## Part (d): Showing $C$ is Compact

To show that $C$ is compact, we must show it is both **closed** and **bounded** (**[Definition 5.43](imo25.pdf#subsection.5.6.2)**).

1.  **Bounded:**
    We proved that $C$ is bounded in Part (c).

2.  **Closed:**
    The set $C$ is defined by non-strict inequalities involving continuous linear functions:
    $$ C = \{ (x,y,z) \in \mathbb{R}^3 \mid g_i(x,y,z) \le c_i, \text{ for } i=1,2,3,4 \} $$
    Subsets defined by $\le$ inequalities of continuous functions are closed sets (**[Proposition 5.42](imo25.pdf#subsection.5.6.2)** and properties of preimages in **[Proposition 5.51](imo25.pdf#subsection.5.7.1)**). An intersection of closed sets is closed (**[Proposition 5.39](imo25.pdf#subsection.5.6.2)**).

**Conclusion:**
Since $C$ is closed and bounded, $C$ is a **compact subset** of $\mathbb{R}^3$.

## Part (e): Existence of Solution

**Problem:** Does the optimization problem $\min f(x,y,z)$ subject to $(x,y,z) \in C$ have a solution?

**Function Definition:**
$$ f(x,y,z) = 12\cos(\sin(e^{x^2-y^3})+y^{17}) + 15xy - e^{x^2-5x+6} $$

**Justification:**
1.  **Continuity of $f$:**
    The function $f$ is composed of polynomial functions, the exponential function, sine, and cosine. All these standard functions are continuous. According to **[Proposition 5.56](imo25.pdf#subsection.5.7.2)** (composition of continuous functions) and **[Proposition 5.57](imo25.pdf#subsection.5.7.2)** (sum/product of continuous functions), $f$ is a continuous function on $\mathbb{R}^3$.

2.  **Compactness of Domain $C$:**
    As proven in Part (d), the set $C$ is a compact subset of $\mathbb{R}^3$.

3.  **The Extreme Value Theorem:**
    **[Theorem 5.66](imo25.pdf#section.5.8)** (The Extreme Value Theorem) states that a continuous function on a non-empty compact subset attains both a global minimum and a global maximum.

**Answer:**
Yes, the optimization problem has a solution because we are minimizing a **continuous function** over a **compact set**.
